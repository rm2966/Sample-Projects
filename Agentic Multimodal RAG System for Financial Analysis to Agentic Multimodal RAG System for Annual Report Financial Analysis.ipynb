````python

# ============================================================================
# CELL 4: PDF Processing Functions
# ============================================================================

class PDFProcessor:
    """Handles PDF extraction and preprocessing for annual reports"""
    
    @staticmethod
    def extract_text_from_pdf(pdf_path: str) -> Dict[str, Any]:
        """Extract text and metadata from PDF file"""
        try:
            text_pages = []
            metadata = {
                'filename': os.path.basename(pdf_path),
                'path': pdf_path,
                'pages': 0,
                'company': None,
                'year': None
            }
            
            # Extract company and year from filename
            filename = os.path.basename(pdf_path)
            if '-' in filename:
                parts = filename.split('-')
                metadata['company'] = parts[0].upper()
                if len(parts) > 1 and parts[1].isdigit():
                    metadata['year'] = parts[1]
            
            with open(pdf_path, 'rb') as file:
                pdf_reader = PyPDF2.PdfReader(file)
                metadata['pages'] = len(pdf_reader.pages)
                
                for page_num, page in enumerate(pdf_reader.pages):
                    text = page.extract_text()
                    text_pages.append({
                        'page': page_num + 1,
                        'content': text
                    })
            
            return {
                'metadata': metadata,
                'pages': text_pages,
                'full_text': '\n\n'.join([p['content'] for p in text_pages])
            }
            
        except Exception as e:
            logger.error(f"Error processing PDF {pdf_path}: {e}")
            return None
    
    @staticmethod
    def clean_financial_text(text: str) -> str:
        """Clean and normalize financial document text"""
        # Remove excessive whitespace
        text = re.sub(r'\s+', ' ', text)
        # Remove page numbers in common formats
        text = re.sub(r'Page \d+ of \d+', '', text)
        text = re.sub(r'^\d+$', '', text, flags=re.MULTILINE)
        # Clean up table formatting artifacts
        text = re.sub(r'[│├─┤┬┴┼]', '', text)
        return text.strip()

print("✅ PDF Processing functions defined")
````

````python

# ============================================================================
# CELL 5: Enhanced EmbeddingsCache and DocumentService
# ============================================================================

class EmbeddingsCache:
    def __init__(self, cache_file: str):
        self.cache_file = cache_file
        self.cache = self._load_cache()
    
    def _load_cache(self) -> dict:
        if os.path.exists(self.cache_file):
            try:
                with open(self.cache_file, 'rb') as f:
                    return pickle.load(f)
            except Exception as e:
                logger.warning(f"⚠️ Could not load cache: {e}")
        return {}
    
    def save_cache(self):
        try:
            with open(self.cache_file, 'wb') as f:
                pickle.dump(self.cache, f)
        except Exception as e:
            logger.error(f"❌ Could not save cache: {e}")
    
    def get_hash(self, text: str) -> str:
        return hashlib.md5(text.encode('utf-8')).hexdigest()
    
    def get(self, text: str) -> Optional[np.ndarray]:
        key = self.get_hash(text)
        return self.cache.get(key)
    
    def set(self, text: str, embedding: np.ndarray):
        key = self.get_hash(text)
        self.cache[key] = embedding


class AnnualReportDocumentService:
    def __init__(self, api_key: str, reports_path: str):
        self.reports_path = reports_path
        self.cache = EmbeddingsCache(CACHE_FILE)
        self.client = OpenAI(api_key=api_key)
        self.index = None
        self.chunks = []
        self.metadata = []
        self.documents = {}  # Store full documents for reference
        self.total_embedding_tokens = 0
        self.text_splitter = RecursiveCharacterTextSplitter(
            chunk_size=CHUNK_SIZE,
            chunk_overlap=CHUNK_OVERLAP,
            separators=["\n\n", "\n", ". ", " ", ""]
        )

    def index_annual_reports(self):
        """Create vector index from annual report PDFs"""
        pdf_files = [f for f in os.listdir(self.reports_path) if f.endswith('.pdf')]
        
        if not pdf_files:
            logger.error("No PDF files found in the directory")
            return "❌ No PDF files found"
        
        logger.info(f"Found {len(pdf_files)} PDF files to process")
        all_chunks = []
        all_metadata = []
        
        for pdf_file in tqdm(pdf_files, desc="Processing PDFs"):
            pdf_path = os.path.join(self.reports_path, pdf_file)
            
            # Extract text from PDF
            pdf_data = PDFProcessor.extract_text_from_pdf(pdf_path)
            if not pdf_data:
                continue
            
            # Store full document
            self.documents[pdf_file] = pdf_data
            
            # Clean text
            cleaned_text = PDFProcessor.clean_financial_text(pdf_data['full_text'])
            
            # Create chunks with page tracking
            chunks = self.text_splitter.split_text(cleaned_text)
            
            # Create metadata for each chunk
            for i, chunk in enumerate(chunks):
                all_chunks.append(chunk)
                all_metadata.append({
                    'filename': pdf_file,
                    'company': pdf_data['metadata']['company'],
                    'year': pdf_data['metadata']['year'],
                    'chunk_index': i,
                    'total_chunks': len(chunks),
                    'file_path': pdf_path
                })
        
        logger.info(f"Created {len(all_chunks)} chunks from {len(pdf_files)} documents")
        
        # Create embeddings
        logger.info("Generating embeddings...")
        embeddings = []
        batch_size = 20
        
        for i in tqdm(range(0, len(all_chunks), batch_size), desc="Creating embeddings"):
            batch = all_chunks[i:i+batch_size]
            batch_embeddings = self.get_embedding_batch(batch)
            embeddings.extend(batch_embeddings)
        
        # Create FAISS index
        embeddings_array = np.array(embeddings).astype('float32')
        faiss.normalize_L2(embeddings_array)
        
        dimension = embeddings_array.shape[1]
        self.index = faiss.IndexFlatIP(dimension)
        self.index.add(embeddings_array)
        
        self.chunks = all_chunks
        self.metadata = all_metadata
        
        # Save index and metadata
        faiss.write_index(self.index, VECTOR_DB_PATH)
        with open(METADATA_PATH, 'wb') as f:
            pickle.dump({
                'chunks': self.chunks,
                'metadata': self.metadata,
                'documents': self.documents
            }, f)
        
        # Save cache
        self.cache.save_cache()
        
        logger.info(f"✅ Index created and saved with {len(self.chunks)} chunks")
        return f"✅ Indexed {len(pdf_files)} annual reports with {len(self.chunks)} chunks"

    def load_index(self):
        """Load existing index"""
        if os.path.exists(VECTOR_DB_PATH) and os.path.exists(METADATA_PATH):
            self.index = faiss.read_index(VECTOR_DB_PATH)
            with open(METADATA_PATH, 'rb') as f:
                data = pickle.load(f)
                self.chunks = data['chunks']
                self.metadata = data['metadata']
                self.documents = data.get('documents', {})
            logger.info(f"✅ Loaded index with {len(self.chunks)} chunks")
            return f"✅ Loaded index with {len(self.chunks)} chunks"
        return None

    @retry(wait=wait_random_exponential(min=1, max=60), stop=stop_after_attempt(5))
    def get_embedding_batch(self, texts: List[str]) -> List[np.ndarray]:
        response = self.client.embeddings.create(
            model=OPENAI_EMBEDDING_MODEL,
            input=texts
        )
        self.total_embedding_tokens += response.usage.total_tokens
        return [np.array(data.embedding) for data in response.data]

    def retrieve_semantic(self, query: str, top_k: int = TOP_K_RETRIEVAL) -> List[Dict]:
        if self.index is None:
            return []
        
        query_emb = self.get_embedding_batch([query])[0].astype('float32').reshape(1, -1)
        faiss.normalize_L2(query_emb)
        scores, indices = self.index.search(query_emb, top_k)
        
        results = []
        for idx, score in zip(indices[0], scores[0]):
            if idx < len(self.chunks):
                results.append({
                    'content': self.chunks[idx],
                    'metadata': self.metadata[idx],
                    'score': float(score)
                })
        
        return results

print("✅ Document Service configured for Annual Reports")
````

````python

# ============================================================================
# CELL 7: Define Original Financial Analysis Tools
# ============================================================================

from langchain.tools import tool

@tool
def search_financial_metrics(keywords: str, company: str = None) -> str:
    """
    Search for specific financial metrics or terms across annual reports.
    
    Parameters:
    - keywords: Financial terms to search (e.g., "revenue", "net income", "EBITDA")
    - company: Optional - filter by company (GOOGL, NVDIA, MSTR)
    """
    logger.info(f"Searching for financial metrics: {keywords}, company: {company}")
    
    results = doc_service.retrieve_semantic(keywords, top_k=30)
    
    if company:
        results = [r for r in results if r['metadata'].get('company', '').upper() == company.upper()]
    
    if not results:
        return f"No results found for '{keywords}'" + (f" in {company}" if company else "")
    
    output = f"📊 **Financial Metrics Search: '{keywords}'**\n\n"
    
    # Group by company
    by_company = {}
    for result in results[:15]:  # Limit display
        comp = result['metadata'].get('company', 'Unknown')
        if comp not in by_company:
            by_company[comp] = []
        by_company[comp].append(result)
    
    for comp, comp_results in by_company.items():
        year = comp_results[0]['metadata'].get('year', 'Unknown')
        output += f"### {comp} ({year})\n"
        for r in comp_results[:3]:
            content_preview = r['content'][:300].replace('\n', ' ')
            output += f"- {content_preview}...\n\n"
    
    return output

@tool
def compare_companies(metric: str) -> str:
    """
    Compare a specific metric across all companies in the database.
    
    Parameters:
    - metric: The metric to compare (e.g., "revenue growth", "operating margin", "cash flow")
    """
    logger.info(f"Comparing metric across companies: {metric}")
    
    results = doc_service.retrieve_semantic(metric, top_k=50)
    
    output = f"📈 **Comparative Analysis: {metric}**\n\n"
    
    # Group by company
    by_company = {}
    for result in results:
        comp = result['metadata'].get('company', 'Unknown')
        if comp not in by_company:
            by_company[comp] = []
        by_company[comp].append(result['content'][:500])
    
    for company, excerpts in by_company.items():
        output += f"### {company}\n"
        # Show most relevant excerpt
        if excerpts:
            output += f"{excerpts[0]}...\n\n"
    
    return output

@tool
def get_company_summary(company: str) -> str:
    """
    Get a summary of key information from a company's annual report.
    
    Parameters:
    - company: Company ticker (GOOGL, NVDIA, MSTR)
    """
    logger.info(f"Getting summary for company: {company}")
    
    # Search for key sections
    key_sections = ["business overview", "financial highlights", "management discussion", "risk factors"]
    
    output = f"📋 **Company Summary: {company}**\n\n"
    
    for section in key_sections:
        results = doc_service.retrieve_semantic(f"{section} {company}", top_k=3)
        results = [r for r in results if r['metadata'].get('company', '').upper() == company.upper()]
        
        if results:
            output += f"### {section.title()}\n"
            output += f"{results[0]['content'][:400]}...\n\n"
    
    return output

@tool
def extract_financial_tables(company: str, table_type: str = "income statement") -> str:
    """
    Extract and format financial tables from annual reports.
    
    Parameters:
    - company: Company ticker
    - table_type: Type of table (income statement, balance sheet, cash flow)
    """
    logger.info(f"Extracting {table_type} for {company}")
    
    query = f"{company} {table_type} financial statements dollars millions thousands"
    results = doc_service.retrieve_semantic(query, top_k=10)
    
    results = [r for r in results if r['metadata'].get('company', '').upper() == company.upper()]
    
    if not results:
        return f"No {table_type} found for {company}"
    
    output = f"📊 **{table_type.title()} - {company}**\n\n"
    
    # Look for chunks with numbers (likely to be tables)
    for result in results:
        # Check if chunk has financial data patterns
        if re.search(r'\$[\d,]+|[\d,]+\.\d{2}|\d{4}\s+\d{4}', result['content']):
            output += f"{result['content'][:800]}\n\n"
            break
    
    return output

@tool
def search_risk_factors(company: str = None) -> str:
    """
    Search for risk factors mentioned in annual reports.
    
    Parameters:
    - company: Optional - filter by specific company
    """
    logger.info(f"Searching risk factors for company: {company}")
    
    query = "risk factors challenges uncertainties threats"
    results = doc_service.retrieve_semantic(query, top_k=30)
    
    if company:
        results = [r for r in results if r['metadata'].get('company', '').upper() == company.upper()]
    
    output = f"⚠️ **Risk Factors Analysis**\n\n"
    
    # Group by company
    by_company = {}
    for result in results[:15]:
        comp = result['metadata'].get('company', 'Unknown')
        if comp not in by_company:
            by_company[comp] = []
        by_company[comp].append(result['content'][:400])
    
    for comp, risks in by_company.items():
        output += f"### {comp}\n"
        for i, risk in enumerate(risks[:3], 1):
            output += f"{i}. {risk}...\n\n"
    
    return output

@tool
def year_over_year_analysis(company: str, metric: str) -> str:
    """
    Analyze year-over-year changes for a specific metric.
    
    Parameters:
    - company: Company ticker
    - metric: Metric to analyze (revenue, profit, expenses, etc.)
    """
    logger.info(f"YoY analysis for {company}: {metric}")
    
    query = f"{company} {metric} growth increase decrease change year over prior"
    results = doc_service.retrieve_semantic(query, top_k=20)
    
    results = [r for r in results if r['metadata'].get('company', '').upper() == company.upper()]
    
    output = f"📈 **Year-over-Year Analysis: {company} - {metric}**\n\n"
    
    for result in results[:5]:
        if re.search(r'\d+%|\$[\d,]+|increase|decrease|growth', result['content'], re.IGNORECASE):
            year = result['metadata'].get('year', 'Unknown')
            output += f"**{year}:**\n{result['content'][:400]}...\n\n"
    
    return output

print("✅ Financial analysis tools defined!")
````

````python
# ============================================================================
# CELL 8: Define Multimodal Enhancement Tools
# ============================================================================

def parse_financial_table(text: str) -> dict:
    """Parse text content to extract table structure"""
    lines = text.split('\n')
    table_data = []
    headers = []
    
    for line in lines:
        # Detect header patterns (Year, Item descriptions)
        if re.search(r'(2019|2020|2021|2022|2023|2024)', line):
            if not headers:
                headers = re.findall(r'\b\w+\b', line)[:5]  # Limit to 5 columns
        # Detect data rows with financial values
        elif re.search(r'\$[\d,]+|[\d,]+\.\d+', line):
            values = re.findall(r'\$?[\d,]+\.?\d*', line)
            if values:
                label = re.sub(r'\$?[\d,]+\.?\d*', '', line).strip()
                if label:
                    table_data.append([label] + values[:4])
    
    if table_data:
        return {'headers': headers or ['Item', 'Value 1', 'Value 2', 'Value 3'], 'data': table_data}
    return None

@tool
def extract_and_format_tables(company: str, keywords: str = "") -> str:
    """
    Extract tables from PDFs and format them for display.
    Searches for financial tables and returns them in markdown format.
    
    Parameters:
    - company: Company ticker (GOOGL, NVDIA, MSTR)
    - keywords: Optional keywords to filter tables (e.g., "revenue", "income statement")
    """
    logger.info(f"Extracting tables for {company} with keywords: {keywords}")
    
    # Search for table-like content
    query = f"{company} {keywords} table financial data numbers millions thousands"
    results = doc_service.retrieve_semantic(query, top_k=10)
    
    # Filter for company
    results = [r for r in results if r['metadata'].get('company', '').upper() == company.upper()]
    
    if not results:
        return f"No tables found for {company} with keywords '{keywords}'"
    
    output = f"## 📊 Financial Tables - {company}\n\n"
    
    # Process each result to identify table-like structures
    tables_found = []
    for result in results[:5]:
        content = result['content']
        # Look for patterns indicating tabular data
        if re.search(r'\$[\d,]+.*\$[\d,]+|\d{4}.*\d{4}.*\d{4}', content):
            # Extract and parse table data
            table_data = parse_financial_table(content)
            if table_data:
                tables_found.append(table_data)
    
    # Format tables using tabulate
    for i, table in enumerate(tables_found[:3], 1):
        output += f"### Table {i}\n"
        output += f"```\n{tabulate.tabulate(table['data'], headers=table['headers'], tablefmt='grid')}\n```\n\n"
    
    return output

@tool
def ranked_search(query: str, search_type: str = "all", top_n: int = 10) -> str:
    """
    Perform ranked search across documents with relevance scoring.
    
    Parameters:
    - query: Search query
    - search_type: "all", "financial_metrics", "risk", "performance"
    - top_n: Number of results to return
    """
    logger.info(f"Performing ranked search: '{query}' (type: {search_type})")
    
    # Get semantic search results with scores
    results = doc_service.retrieve_semantic(query, top_k=top_n * 2)
    
    # Apply type-specific filtering
    if search_type == "financial_metrics":
        results = [r for r in results if re.search(r'\$[\d,]+|revenue|income|expense', r['content'], re.I)]
    elif search_type == "risk":
        results = [r for r in results if re.search(r'risk|uncertainty|challenge|threat', r['content'], re.I)]
    elif search_type == "performance":
        results = [r for r in results if re.search(r'growth|increase|decrease|performance', r['content'], re.I)]
    
    # Sort by relevance score
    results = sorted(results, key=lambda x: x['score'], reverse=True)[:top_n]
    
    # Format as ranked list with scores
    output = f"## 🔍 Ranked Search Results for: '{query}'\n\n"
    output += f"**Search Type:** {search_type} | **Results:** {len(results)}\n\n"
    
    # Create summary table
    summary_data = []
    for i, result in enumerate(results, 1):
        company = result['metadata'].get('company', 'Unknown')
        year = result['metadata'].get('year', 'Unknown')
        score = result['score']
        preview = result['content'][:100] + "..."
        summary_data.append([i, company, year, f"{score:.3f}", preview])
    
    # Display as table
    headers = ["Rank", "Company", "Year", "Score", "Preview"]
    output += f"```\n{tabulate.tabulate(summary_data, headers=headers, tablefmt='grid')}\n```\n\n"
    
    # Detailed results
    output += "### Detailed Results:\n\n"
    for i, result in enumerate(results[:5], 1):
        output += f"**Rank {i} (Score: {result['score']:.3f})**\n"
        output += f"- Company: {result['metadata'].get('company', 'Unknown')}\n"
        output += f"- Page: {result['metadata'].get('chunk_index', 'Unknown')}\n"
        output += f"- Content: {result['content'][:300]}...\n\n"
    
    return output

@tool
def create_comparison_chart(metric: str, output_format: str = "table") -> str:
    """
    Create visual comparisons of metrics across companies.
    
    Parameters:
    - metric: The metric to compare (e.g., "revenue", "operating margin")
    - output_format: "table", "chart_description", or "both"
    """
    logger.info(f"Creating comparison for metric: {metric}")
    
    # Gather data for all companies
    companies = ["GOOGL", "NVDIA", "MSTR"]
    comparison_data = []
    
    for company in companies:
        results = doc_service.retrieve_semantic(f"{company} {metric} 2023 2024", top_k=5)
        company_results = [r for r in results if r['metadata'].get('company', '').upper() == company.upper()]
        
        if company_results:
            # Extract numerical values
            content = company_results[0]['content']
            values = re.findall(r'\$?([\d,]+\.?\d*)\s*(?:billion|million)?', content)
            if values:
                # Convert to float (assuming millions)
                try:
                    value = float(values[0].replace(',', ''))
                    comparison_data.append({
                        'Company': company,
                        'Metric': metric.title(),
                        'Value': value,
                        'Context': content[:200]
                    })
                except:
                    pass
    
    if not comparison_data:
        return f"No comparable data found for '{metric}'"
    
    output = f"## 📈 Comparative Analysis: {metric.title()}\n\n"
    
    # Create table view
    if output_format in ["table", "both"]:
        table_data = [[d['Company'], f"${d['Value']:,.0f}", d['Context'][:50]+"..."] 
                      for d in comparison_data]
        headers = ["Company", metric.title(), "Context"]
        output += "### Tabular Comparison\n"
        output += f"```\n{tabulate.tabulate(table_data, headers=headers, tablefmt='grid')}\n```\n\n"
    
    # Create chart description (in lieu of actual chart)
    if output_format in ["chart_description", "both"]:
        output += "### Visual Analysis\n"
        sorted_data = sorted(comparison_data, key=lambda x: x['Value'], reverse=True)
        
        output += f"**Ranking by {metric.title()}:**\n"
        for i, d in enumerate(sorted_data, 1):
            bar_length = int(d['Value'] / max(c['Value'] for c in comparison_data) * 30)
            bar = "█" * bar_length
            output += f"{i}. {d['Company']}: {bar} ${d['Value']:,.0f}\n"
        
        # Add insights
        output += f"\n**Key Insights:**\n"
        output += f"- Highest {metric}: {sorted_data[0]['Company']} (${sorted_data[0]['Value']:,.0f})\n"
        output += f"- Lowest {metric}: {sorted_data[-1]['Company']} (${sorted_data[-1]['Value']:,.0f})\n"
        if len(sorted_data) > 1:
            spread = sorted_data[0]['Value'] - sorted_data[-1]['Value']
            output += f"- Spread: ${spread:,.0f}\n"
    
    return output

@tool
def aggregate_financial_data(metrics: list, years: list = None) -> str:
    """
    Aggregate financial data across all companies into a unified table.
    
    Parameters:
    - metrics: List of metrics to aggregate ["revenue", "net income", "operating margin"]
    - years: Optional list of years to include
    """
    logger.info(f"Aggregating financial data for metrics: {metrics}")
    
    if not years:
        years = ["2023", "2024"]
    
    companies = ["GOOGL", "NVDIA", "MSTR"]
    aggregated_data = {}
    
    for company in companies:
        aggregated_data[company] = {}
        for metric in metrics:
            query = f"{company} {metric} " + " ".join(years)
            results = doc_service.retrieve_semantic(query, top_k=3)
            
            # Filter for company
            company_results = [r for r in results 
                              if r['metadata'].get('company', '').upper() == company.upper()]
            
            if company_results:
                # Extract value
                content = company_results[0]['content']
                values = re.findall(r'\$?([\d,]+\.?\d*)\s*(?:billion|million)?', content)
                if values:
                    aggregated_data[company][metric] = values[0]
                else:
                    aggregated_data[company][metric] = "N/A"
            else:
                aggregated_data[company][metric] = "N/A"
    
    # Format as comprehensive table
    output = f"## 📊 Aggregated Financial Data\n\n"
    output += f"**Metrics:** {', '.join(metrics)}\n"
    output += f"**Years:** {', '.join(years)}\n\n"
    
    # Create table
    table_data = []
    for company in companies:
        row = [company]
        for metric in metrics:
            value = aggregated_data[company].get(metric, "N/A")
            row.append(value)
        table_data.append(row)
    
    headers = ["Company"] + [m.title() for m in metrics]
    output += f"```\n{tabulate.tabulate(table_data, headers=headers, tablefmt='grid')}\n```\n\n"
    
    # Add summary statistics
    output += "### Summary Statistics\n"
    for metric in metrics:
        output += f"\n**{metric.title()}:**\n"
        for company in companies:
            value = aggregated_data[company].get(metric, "N/A")
            output += f"- {company}: {value}\n"
    
    return output

print("✅ Multimodal enhancement tools defined!")
````

````python

# ============================================================================
# CELL 10: Gradio Interface for Enhanced Financial Analysis
# ============================================================================

def create_gradio_interface():
    def chat_function(message, history):
        if not agent_executor:
            return "⚠️ Agent not initialized. Please check the index."
        
        chat_history = []
        if history:
            for user_msg, ai_msg in history:
                if user_msg:
                    chat_history.append(HumanMessage(content=user_msg))
                if ai_msg:
                    chat_history.append(AIMessage(content=ai_msg))
        
        try:
            with get_openai_callback() as cb:
                response = agent_executor.invoke({
                    "input": message,
                    "chat_history": chat_history
                })
                answer = response.get('output', 'Sorry, I could not process that request.')
                
                # Cost calculation
                input_cost = (cb.prompt_tokens / 1_000_000) * 0.15
                output_cost = (cb.completion_tokens / 1_000_000) * 0.60
                total_cost = input_cost + output_cost
                
                cost_info = f"\n\n---\n💰 **Usage:** ${total_cost:.6f} | Tokens: {cb.total_tokens:,}"
                
                # Save query log
                timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
                log_file = os.path.join(OUTPUT_DIRECTORY, f"query_{timestamp}.md")
                with open(log_file, 'w', encoding='utf-8') as f:
                    f.write(f"# Query: {message}\n\n{answer}\n\n{cost_info}")
                
                return answer + cost_info
                
        except Exception as e:
            logger.error(f"Error: {e}")
            return f"❌ Error: {str(e)}"
    
    interface = gr.ChatInterface(
        fn=chat_function,
        title="📊 Multimodal Annual Reports Financial Analysis System",
        description=f"""
        **Enhanced RAG System with Table Extraction and Ranked Search**
        
        Analyze annual reports from GOOGL (2024), NVDIA (2024), and MSTR (2023)
        
        **New Features:**
        - 📊 Table extraction and formatting
        - 🔍 Ranked search with relevance scores
        - 📈 Visual comparisons and charts
        - 📑 Cross-document data aggregation
        
        **Sample Queries:**
        - "Show me ranked results for revenue growth"
        - "Extract financial tables for NVIDIA"
        - "Create a comparison chart for operating margins"
        - "Aggregate revenue and net income data for all companies"
        
        Results saved to: `{OUTPUT_DIRECTORY}`
        """,
        examples=[
            "Compare the revenue of all three companies",
            "Show me ranked results for AI investments",
            "Extract income statement tables for GOOGL",
            "Create a comparison chart for R&D expenses",
            "What are NVDIA's main risk factors?",
            "Aggregate key financial metrics across all companies"
        ],
        cache_examples=False
    )
    
    return interface

print("✅ Enhanced Gradio interface configured")
````

````python
# ============================================================================
# CELL 12: Simple Image Extraction from NVIDIA PDF - No Extra Dependencies
# ============================================================================

import fitz  # PyMuPDF - you already have this
from PIL import Image
import io
import os

def extract_nvidia_shareholder_charts():
    """
    Extract images and charts from NVIDIA's annual report
    focusing on shareholder returns section
    """
    
    # Find NVIDIA PDF
    nvidia_pdf = None
    for pdf_file in os.listdir(ANNUAL_REPORTS_PATH):
        if 'NVDIA' in pdf_file.upper() or 'NVIDIA' in pdf_file.upper():
            nvidia_pdf = os.path.join(ANNUAL_REPORTS_PATH, pdf_file)
            break
    
    if not nvidia_pdf:
        print("❌ NVIDIA PDF not found in directory")
        return None
    
    print(f"📄 Found NVIDIA PDF: {nvidia_pdf}")
    
    # Create output directories
    images_dir = os.path.join(OUTPUT_DIRECTORY, "nvidia_images")
    pages_dir = os.path.join(OUTPUT_DIRECTORY, "nvidia_pages")
    os.makedirs(images_dir, exist_ok=True)
    os.makedirs(pages_dir, exist_ok=True)
    
    # Open PDF
    pdf_document = fitz.open(nvidia_pdf)
    
    # Step 1: Find pages containing shareholder returns content
    print("\n🔍 Searching for Shareholder Returns pages...")
    target_pages = []
    
    for page_num in range(len(pdf_document)):
        page = pdf_document[page_num]
        text = page.get_text().upper()
        
        # Keywords that indicate shareholder returns chart
        keywords = ['SHAREHOLDER RETURN', 'TSR', '1436%', '371%', '200%',
                   'RETURN OF CAPITAL', 'STOCK REPURCHASES', '$21.0', '$19.8',
                   'FISCAL 2024 SHAREHOLDER', 'DIVIDENDS']
        
        matching_keywords = [kw for kw in keywords if kw in text]
        
        if matching_keywords:
            target_pages.append(page_num)
            print(f"  ✓ Page {page_num + 1}: Found {matching_keywords[:3]}...")
    
    if not target_pages:
        print("⚠️ No pages found with shareholder returns keywords")
        # Fall back to checking all pages with images
        for page_num in range(len(pdf_document)):
            page = pdf_document[page_num]
            if page.get_images():
                target_pages.append(page_num)
    
    print(f"\n📊 Processing {len(target_pages)} relevant pages...")
    
    # Step 2: Extract images from target pages
    extracted_items = []
    
    for page_num in target_pages[:10]:  # Limit to first 10 relevant pages
        page = pdf_document[page_num]
        
        # A. Extract embedded images
        image_list = page.get_images(full=True)
        
        if image_list:
            print(f"\nPage {page_num + 1}: Found {len(image_list)} embedded images")
            
            for img_index, img in enumerate(image_list):
                try:
                    # Extract image
                    xref = img[0]
                    base_image = pdf_document.extract_image(xref)
                    image_bytes = base_image["image"]
                    
                    # Save image
                    image_filename = f"page_{page_num + 1}_image_{img_index + 1}.{base_image['ext']}"
                    image_path = os.path.join(images_dir, image_filename)
                    
                    with open(image_path, "wb") as f:
                        f.write(image_bytes)
                    
                    # Get dimensions
                    pil_img = Image.open(io.BytesIO(image_bytes))
                    
                    extracted_items.append({
                        'type': 'embedded_image',
                        'page': page_num + 1,
                        'path': image_path,
                        'size': pil_img.size
                    })
                    
                    print(f"  ✓ Saved: {image_filename} ({pil_img.size[0]}x{pil_img.size[1]})")
                    
                except Exception as e:
                    print(f"  ⚠️ Failed to extract image {img_index + 1}: {e}")
        
        # B. Render entire page as image (captures vector graphics/charts)
        print(f"  📸 Rendering page {page_num + 1} as image...")
        
        # High quality render
        mat = fitz.Matrix(200/72.0, 200/72.0)  # 200 DPI
        pix = page.get_pixmap(matrix=mat)
        
        page_filename = f"page_{page_num + 1}_full.png"
        page_path = os.path.join(pages_dir, page_filename)
        pix.save(page_path)
        
        extracted_items.append({
            'type': 'page_render',
            'page': page_num + 1,
            'path': page_path,
            'size': (pix.width, pix.height)
        })
        
        print(f"  ✓ Saved full page: {page_filename}")
    
    pdf_document.close()
    
    # Step 3: Summary
    print("\n" + "="*60)
    print(f"✅ EXTRACTION COMPLETE")
    print(f"   Total items extracted: {len(extracted_items)}")
    print(f"   Images saved to: {images_dir}")
    print(f"   Page renders saved to: {pages_dir}")
    print("="*60)
    
    return extracted_items

def display_extraction_summary(extracted_items):
    """
    Display a summary of extracted items
    """
    if not extracted_items:
        return
    
    print("\n📋 EXTRACTION SUMMARY:")
    print("-" * 50)
    
    # Group by type
    images = [i for i in extracted_items if i['type'] == 'embedded_image']
    pages = [i for i in extracted_items if i['type'] == 'page_render']
    
    if images:
        print(f"\n🖼️ Embedded Images ({len(images)}):")
        for img in images[:5]:  # Show first 5
            print(f"   • Page {img['page']}: {img['size'][0]}x{img['size'][1]} pixels")
            print(f"     {img['path']}")
    
    if pages:
        print(f"\n📄 Page Renders ({len(pages)}):")
        for page in pages[:5]:  # Show first 5
            print(f"   • Page {page['page']}: {page['size'][0]}x{page['size'][1]} pixels")
            print(f"     {page['path']}")
    
    print("\n💡 TIP: The shareholder returns chart should be visible in:")
    print("   1. Check the page renders first (full page screenshots)")
    print("   2. Look for pages mentioning '1436%', '371%', or 'TSR'")
    print("   3. The chart has green bars showing these percentages")

# ============================================================================
# Run the extraction
# ============================================================================

print("="*70)
print("EXTRACTING NVIDIA SHAREHOLDER RETURNS IMAGES")
print("="*70)

# Execute extraction
extracted = extract_nvidia_shareholder_charts()

# Show summary
if extracted:
    display_extraction_summary(extracted)
    
    # Also search in the text for the exact page
    print("\n🔍 Based on text analysis, the shareholder returns chart")
    print("   is most likely in the page renders for pages that")
    print("   contain '1436%', 'TSR', or 'Return of Capital'")
else:
    print("\n❌ No images extracted. Check if the PDF exists and is readable.")
````
